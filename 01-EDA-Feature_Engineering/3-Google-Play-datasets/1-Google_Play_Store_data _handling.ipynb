### Data Cleaning

##### Rating

```python
df_copy['rating'].isnull().sum()

df_copy['rating'].dtype
```

##### Reviews

```python
df_copy['reviews'].isnull().sum()

df_copy[~df_copy['reviews'].str.isnumeric()]

df_copy['reviews'] = df_copy['reviews'].str.replace('M', '')

df_copy['reviews']=df_copy['reviews'].astype(float)

df_copy['reviews'].dtype
```

##### Size to size_in_k everything is exist in Thousand form

```python
#### Size to size_in_k everything is exist in Thousand form
df_copy['size'] = df['Size'].copy()
```

```python
df_copy['size'].isnull().sum()
```

```python
df_copy['size'].unique()
```

```python
df_copy['size'] = df_copy['size'].str.replace('M', '00k')

df_copy['size'] = df_copy['size'].replace('Varies with device', np.nan)

df_copy['size'].unique()

df_copy['size_in_k'] = df_copy['size'].str.replace('k', '')

df_copy['size_in_k'] = df_copy['size_in_k'].str.replace(',', '')
df_copy['size_in_k'] = df_copy['size_in_k'].str.replace('+', '')
df_copy['size_in_k'] = df_copy['size_in_k'].str.replace('.', '')

df_copy['size_in_k'].dtype
df_copy.drop('size', axis=1, inplace=True)

df_copy['size_in_k'] = df_copy['size_in_k'].astype(float)
```

```python
df_copy.info()
```

#### Installs

```python
df_copy['installs'].isnull().sum()
```

```python
df_copy['installs'].unique()
```

```python
df_copy['installs'] = df_copy['installs'].str.replace('+', '')
df_copy['installs'] = df_copy['installs'].str.replace(',', '')
df_copy['installs'] = df_copy['installs'].replace('Free', np.nan)
```

```python
df_copy['installs'] = df_copy['installs'].astype(float)
```

```python
df_copy.info()
```

#### Type

```python
df_copy['type'].value_counts()
```

```python
df_copy['type'].isnull().sum()
```

```python
df_copy['type'] = df_copy['type'].fillna(df_copy['type'].mode()[0])

df_copy['type'] = df_copy['type'].replace({'0' : df_copy['type'].mode()[0]})

df_copy['type'] = df_copy['type'].map({'Free' : 0, 'Paid' :1})

df_copy['type'] = df_copy['type'].astype(int)
```

```python
df_copy.info()
```

#### Price

```python
df_copy['price'].unique()
```

```python
df_copy['price'] = df_copy['price'].str.replace('$', '')
```

```python
df_copy['price'] = df_copy['price'].str.replace('Everyone', '0')
```

```python
df_copy['price'] = df_copy['price'].astype(float)
```

```python
df_copy.info()
```

#### Content Rating

```python
df_copy['content rating'].isnull().sum()
```

```python
df_copy['content rating'] = df_copy['content rating'].fillna(df_copy['content rating'].mode()[0])
```

```python
df_copy['content rating'].unique()
```
