```python
x = df.drop(columns='selling_price', axis=1)
y = df['selling_price']
```

# Data Encoding

```python
# Label Encoding for df.model bcz -> it have 120 different categories
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
x['model'] = le.fit_transform(x['model'])
```

```python
categorical_feature = df.select_dtypes(include='object').columns
numerical_feature = x.select_dtypes(exclude='object').columns

from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

Oh = OneHotEncoder(drop='first', sparse_output=False)
stdscl = StandardScaler()

processor = ColumnTransformer([
    ("OneHotEncoder", Oh, categorical_feature),
    ("StandardScaler", stdscl, numerical_feature)
], remainder='passthrough')
```

```python
x = processor.fit_transform(x)
```

```python
pd.DataFrame(x)
```

# Train Test split - 

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

```python
x_train
```

```python
def metric_eval(true, pred):
    mae = mean_absolute_error(true, pred)
    mse = mean_squared_error(true, pred)
    r2 = r2_score(true, pred)
    rmse = np.sqrt(mse)
    return mae, mse, rmse, r2
```

```python
# Multiple model comparision -
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
models = {
    "Linear" : LinearRegression(),
    "Ridge" : Ridge(),
    "Lasso" : Lasso(),
    "Decision Tree" : DecisionTreeRegressor()
}

for i in range(len(models)):
    model = list(models.values())[i]
    model.fit(x_train, y_train)
    train_pred = model.predict(x_train)
    test_pred = model.predict(x_test)

    tr_mae, tr_mse, tr_rmse, tr_r2 =metric_eval(y_train, train_pred)
    t_mae, t_mse, t_rmse, t_r2 =metric_eval(y_test, test_pred)

    print(f"For {list(models.keys())[i]} model -----\n")
    print('Model performance for Training set')
    print("- Mean Squared Error: {:.4f}".format(tr_mse))
    print("- Root Mean Squared Error: {:.4f}".format(tr_rmse))
    print("- Mean Absolute Error: {:.4f}".format(tr_mae))
    print("- R2 Score: {:.4f}".format(tr_r2))

    print('----------------------------------')
    
    print('Model performance for Test set')
    print("- Mean Squared Error: {:.4f}".format(t_mse))
    print("- Root Mean Squared Error: {:.4f}".format(t_rmse))
    print("- Mean Absolute Error: {:.4f}".format(t_mae))
    print("- R2 Score: {:.4f}".format(t_r2))
    
    print('='*35)
    print('\n')

    
    
```

# Hyperparameter Tuning --

#### GridSearchCV

```python
# Hyperparameters for GridSearchCV
from sklearn.model_selection import GridSearchCV
param_for_grid = {
    'max_depth': [5, 10, 15, 20, 25, None],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 5, 10]
}

grid_cv_model = [
    ("Decision Tree", DecisionTreeRegressor(), param_for_grid)
]
```

```python
model_best_param_Grid = {}

for name, model, param in grid_cv_model:
    GridCV = GridSearchCV(
        estimator=model,
        param_grid=param,
        cv=7,
        n_jobs=-1,
        scoring="r2",
    )
    GridCV.fit(x_train, y_train)
    model_best_param_Grid[name] = [GridCV.best_params_, GridCV.best_score_]

for val in model_best_param_Grid:
    print(model_best_param_Grid[val][0])
    print(model_best_param_Grid[val][1])
```

```python
models = {
    "Decision Tree" : DecisionTreeRegressor(
        min_samples_split=10, min_samples_leaf=1, max_depth=15
    )
}


for i in range(len(models)):
    model = list(models.values())[i]
    model.fit(x_train, y_train)

    train_pred = model.predict(x_train)
    test_pred = model.predict(x_test)

    tr_mae, tr_mse, tr_rmse, tr_r2 =metric_eval(y_train, train_pred)
    t_mae, t_mse, t_rmse, t_r2 =metric_eval(y_test, test_pred)

    print(f"For {list(models.keys())[i]} model -----\n")
    print('Model performance for Training set')
    print("- Mean Squared Error: {:.4f}".format(tr_mse))
    print("- Root Mean Squared Error: {:.4f}".format(tr_rmse))
    print("- Mean Absolute Error: {:.4f}".format(tr_mae))
    print("- R2 Score: {:.4f}".format(tr_r2))

    print('----------------------------------')
    
    print('Model performance for Test set')
    print("- Mean Squared Error: {:.4f}".format(t_mse))
    print("- Root Mean Squared Error: {:.4f}".format(t_rmse))
    print("- Mean Absolute Error: {:.4f}".format(t_mae))
    print("- R2 Score: {:.4f}".format(t_r2))
    
    print('='*35)
    print('\n')
```

#### RandomSearchCV

```python
from sklearn.model_selection import RandomizedSearchCV
param_for_rand = {
    "criterion" : ['squared_error', 'absolute_error', 'poisson'],
    "splitter" : ['best', 'random'],
    "max_depth" : [10, 20, 40, 60, 80, 100],
    "min_samples_split" : [2, 4, 8, 10, 20],
    "min_samples_leaf" : [1, 2, 3, 4, 5],
    "max_features": [5, 7, 8, 10],
}

random_cv_model = [
    ("Decision Tree", DecisionTreeRegressor(), param_for_rand)
]
```

```python
model_best_param_rand ={}

for name, model, param in random_cv_model:
    RandSrchCV = RandomizedSearchCV(estimator=model,
                                    param_distributions=param,
                                    scoring="r2",
                                    cv=3,
                                    n_jobs=-1,
                                    verbose=2,
                                    n_iter=100,
                                    random_state=42
                                   )
    RandSrchCV.fit(x_train, y_train)
    model_best_param_rand[model] = [RandSrchCV.best_params_, RandSrchCV.best_score_]
    
for val in model_best_param_rand:
    print(model_best_param_rand[val][0])
    print(model_best_param_rand[val][1])
```
