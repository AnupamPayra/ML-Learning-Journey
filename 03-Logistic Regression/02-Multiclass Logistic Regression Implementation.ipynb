```python
import numpy as np
import pandas as pd
```

### Creating dataset --

```python
from sklearn.datasets import make_classification

x, y = make_classification(n_samples=1000, n_features=10, random_state=42, n_informative=3,  n_classes=3)
```

```python
# train test split --
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=.30, random_state=42)
```

### Model Implementing --

```python
from sklearn.linear_model import LogisticRegression
```

```python
regression = LogisticRegression(multi_class='ovr')
```

```python
regression.fit(x_train, y_train)
```

```python
y_pred = regression.predict(x_test)
```

```python
# Model evaluation ---
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print(confusion_matrix(y_test, y_pred))
print('Accurecy - ', accuracy_score(y_test, y_pred))
print("+++++++++++++++++++++++++++++++++++++++++++++++++++++++++")
print(classification_report(y_test, y_pred))
```

### Hyperparameter Tuning---
#### GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
```

```python
model = LogisticRegression()
param_grid = [
    {
        'penalty': ['l1'],
        'solver': ['liblinear', 'saga'],
        'C': [0.01, 0.1, 1, 10, 100]
    },
    {
        'penalty': ['l2'],
        'solver': ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga'],
        'C': [0.01, 0.1, 1, 10, 100]
    },
    {
        'penalty': ['elasticnet'],
        'solver': ['saga'],
        'C': [0.01, 0.1, 1, 10, 100],
        'l1_ratio' : [0.001, 0.004, 0.006, 0.02, 0.1]
    }
]
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold(n_splits=5)
```

```python
grid = GridSearchCV(estimator=model, param_grid=param_grid, n_jobs=-1, cv=cv, scoring='accuracy')
```

```python
grid.fit(x_train, y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
grid_y_pred = grid.predict(x_test)
```

```python
print(confusion_matrix(y_test, grid_y_pred))
print('Accurecy - ', accuracy_score(y_test, grid_y_pred))
print("+++++++++++++++++++++++++++++++++++++++++++++++++++++++++")
print(classification_report(y_test, grid_y_pred))
```

### Hyperparameter Tuning---
#### RandomizedCV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
param_grid = [
    {
        'penalty': ['l1'],
        'solver': ['liblinear', 'saga'],
        'C': [0.01, 0.1, 1, 10, 100]
    },
    {
        'penalty': ['l2'],
        'solver': ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga'],
        'C': [0.01, 0.1, 1, 10, 100]
    },
    {
        'penalty': ['elasticnet'],
        'solver': ['saga'],
        'C': [0.01, 0.1, 1, 10, 100],
        'l1_ratio' : [0.001, 0.004, 0.006, 0.02, 0.1]
    }
]
```

```python
randcv = RandomizedSearchCV(estimator=model, param_distributions=param_grid, cv=cv, scoring='accuracy', n_jobs=-1)
```

```python
randcv.fit(x_train, y_train)
```

```python
print(randcv.best_params_)
print(randcv.best_score_)
```

```python
print(confusion_matrix(y_test, rand_y_pred))
print('Accurecy - ', accuracy_score(y_test, rand_y_pred))
print("+++++++++++++++++++++++++++++++++++++++++++++++++++++++++")
print(classification_report(y_test, rand_y_pred))
```
