```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Clean Google Play data.csv')
```

```python
df.head(2)
```

```python
df.drop('Unnamed: 0', axis=1, inplace=True)
```

```python
df_copy = df.copy()
```

```python
df_copy['app'].duplicated().sum()
```

```python
## Observation = dataset have 1181 duplicate values
```

```python
df_copy.drop_duplicates(subset=['app'], keep='first', inplace=True)
```

```python
df_copy.shape
```

```python
### Exploring Data
```

```python
numerical_features = [x for x in df_copy.columns if df_copy[x].dtype != 'O']
categorical_features = [x for x in df_copy.columns if df_copy[x].dtype == 'object']
```

```python
numerical_features
```

```python
categorical_features
```

```python
### Checking categorical data's distribution ---
for feature in categorical_features:
    sns.barplot(x=feature, data=df_copy)
    plt.show()
    
```
